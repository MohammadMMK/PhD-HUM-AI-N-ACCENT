```python
from pathlib import Path
from markitdown import MarkItDown

# Change these paths
INPUT_DIR = Path("C:\\Users\\utente\\Desktop\\New folder\\")
OUTPUT_DIR = Path("C:\\Users\\utente\\Desktop\\New folder\\")

converter = MarkItDown()
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Recursively process every file in INPUT_DIR
for source_file in INPUT_DIR.rglob("*"):
    if not source_file.is_file():
        continue

    # Preserve the directory structure and avoid overwriting files
    relative_path = source_file.relative_to(INPUT_DIR)
    output_file = OUTPUT_DIR / f"{relative_path}.md"
    output_file.parent.mkdir(parents=True, exist_ok=True)

    try:
        result = converter.convert(str(source_file))

        output_file.write_text(
            result.text_content,
            encoding="utf-8"
        )

        print(f"Converted: {source_file} -> {output_file}")

    except Exception as error:
        print(f"Skipped: {source_file}")
        print(f"Reason: {error}")
```
