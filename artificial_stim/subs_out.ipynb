```python
import time
import torch
import soundfile as sf
from pathlib import Path
from functions import (
    text_to_ipa_chunks, manipulate, check_ipa,
    synth_aligned, audiovisualize_interactive,
)

VOICES_DIR = Path("custom_voices")
OUT_DIR = Path("output_online"); OUT_DIR.mkdir(exist_ok=True)
SKIP_EXISTING = False      # don't regenerate versions that are already in outputs/

voice     = torch.load(VOICES_DIR / "nicola_80.pt",  weights_only=True)   # the voice you hear

text = """La casa di Carlo è dietro il parco; Il cane corre sul campo.
Chiara compra un chilo di pane; Questo libro costa poco.
Marco beve un caffè caldo; La scuola è chiusa per il cattivo tempo.
Il cuoco prepara la carne; Luca ha comprato una borsa nera per Carla.
Che bella giornata per andare in campagna con Paolo; Quanto costa il biglietto per Como?"""

ipa_chunks = text_to_ipa_chunks(text)
check_ipa(ipa_chunks)
r_subs = ["h", "x", "χ", "ɰ", 'ʔ', "q", "kʰ", "kʲ"]

all_chunks, all_masks, all_rules, all_stats = [], [], [], {}   # NEW: collect every version
for alt in r_subs:

    rules = [
            {"kind": "substitute_prob", "old": "k", "choices": {alt: 100}, "same_in_word": True},]

    manip_chunks, masks, stats, origins = manipulate(
                ipa_chunks, rules, return_mask=True, return_stats=True, return_origin=True, seed=42)

    all_chunks += list(manip_chunks)   # NEW: append this version after the previous ones
    all_masks  += list(masks)
    all_rules  += rules
    all_stats[alt] = stats

name = 'test3'
full_text = "\n".join([text] * len(r_subs))   # NEW: text repeated once per substitution

audio, segments, seg_words, seg_altered = synth_aligned(all_chunks, voice, speed=1, masks=all_masks)
audiovisualize_interactive(
    audio, segments, seg_words=seg_words, text=full_text, seg_altered=seg_altered,
    title=name, rules=all_rules, stats=all_stats, out_html=f"{name}.html")
sf.write(OUT_DIR / f"{name}.wav", audio, 24000)
```
